```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv("wbc.csv")
df.head()
```

```python
print(df.columns.tolist())
df.info()
```

```python
# Cek kolom yang perlu dibuang
kolom_dibuang = [col for col in df.columns if col.lower() == 'id' or 'unnamed' in col.lower()]
print("Kolom yang dibuang:", kolom_dibuang)

df = df.drop(columns=kolom_dibuang)
```

```python
df.head()
```

```python
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])

print("Mapping label:", dict(zip(le.classes_, le.transform(le.classes_))))
df['diagnosis'].value_counts()
```

```python
kolom_numerik = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
kolom_numerik.remove('diagnosis')  # kolom target, jangan ikut distandardisasi

scaler = StandardScaler()
df[kolom_numerik] = scaler.fit_transform(df[kolom_numerik])
```
