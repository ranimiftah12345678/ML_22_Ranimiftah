## Kegunaan Masing-Masing Library

**1. PyPREP**
Library untuk membersihkan data sinyal otak (EEG). PyPREP membantu membuang noise/gangguan, mendeteksi sensor (channel) yang rusak, dan merapikan data EEG sebelum dipakai untuk analisis atau melatih model machine learning.

**2. Scipy**
Library "kotak perkakas" matematika dan sains untuk Python. Berisi fungsi-fungsi siap pakai untuk statistik, aljabar linear, optimasi, dan pemrosesan sinyal. Banyak library lain (termasuk PyPREP) bergantung pada Scipy sebagai fondasi perhitungannya.

**3. wandb (Weights & Biases)**
Library untuk mencatat dan memantau proses eksperimen machine learning. Dengan wandb, setiap kali kita mencoba model dengan setting berbeda (misalnya learning rate), hasilnya (akurasi, loss, dsb) otomatis tersimpan dan bisa dibandingkan lewat dashboard — seperti buku catatan digital untuk eksperimen ML.

**4. pyECG**
Library untuk membaca dan menulis data sinyal jantung (ECG/EKG) dari format file umum seperti WFDB dan iShine. Berguna untuk memuat file rekaman detak jantung pasien ke Python agar bisa dianalisis atau dipakai melatih model.

### 1. Contoh Pelanggaran Etika dan Hukum Penggunaan AI

Salah satu kasus nyata di Indonesia adalah penyalahgunaan teknologi **deepfake** — AI yang dipakai untuk memalsukan wajah, suara, atau video seseorang. Salah satu contoh yang ramai diberitakan adalah kasus Baim Wong, di mana teknologi deepfake dan kloning suara digunakan bukan untuk inovasi, melainkan untuk manipulasi dan penipuan online yang merugikan korban secara materi maupun psikologis. Contoh lain, ada juga kasus penipuan yang memakai teknologi AI deepfake dengan wajah Presiden Prabowo Subianto untuk mengelabui korban.

Secara hukum, Indonesia belum punya undang-undang khusus yang mengatur etika dan penggunaan AI, sehingga UU ITE (Undang-Undang Informasi dan Transaksi Elektronik) masih menjadi dasar hukum utama untuk menjerat pelaku penyalahgunaan AI seperti deepfake. Secara spesifik, Pasal 27A dan Pasal 29 UU ITE bisa dipakai untuk menjerat pelaku penyebaran konten deepfake yang memuat penghinaan, ancaman, atau pencemaran nama baik.

Kasus ini menunjukkan pelanggaran etika AI dalam bentuk: penggunaan data pribadi/wajah orang tanpa izin, penipuan, dan penyebaran informasi palsu — semuanya melanggar prinsip dasar etika AI seperti *consent* (persetujuan), kejujuran, dan tanggung jawab.

### 2. Dampak Energi dan Lingkungan dari AI, serta Cara Mengatasinya

AI, terutama AI generatif (seperti ChatGPT), membutuhkan pusat data (data center) raksasa yang menyala nonstop untuk melatih dan menjalankan model. Ini menimbulkan dua dampak utama:

**a. Konsumsi listrik dan emisi karbon meningkat tajam.**
International Energy Agency (IEA) memproyeksikan konsumsi listrik pusat data global akan meningkat lebih dari dua kali lipat, mencapai sekitar 945 TWh pada 2030, atau sekitar 3% dari konsumsi listrik global. Untuk Indonesia sendiri, emisi karbon pusat data berpotensi meningkat hampir empat kali lipat dari 2024 ke 2030.

**b. Konsumsi air yang sangat besar.**
Pusat data AI membutuhkan pendinginan agar server tidak overheat. MIT News melaporkan bahwa pusat data yang menjalankan AI generatif berpotensi menggunakan hingga jutaan liter air setiap tahun hanya untuk menjaga suhu server tetap stabil.

**Cara mengatasi (pendapat pribadi):**
Menurut saya, solusi paling realistis bukan berhenti memakai AI, tapi membuat penggunaannya lebih efisien dan bertanggung jawab, misalnya:
- Mengadopsi energi terbarukan seperti surya dan angin untuk menekan emisi karbon, serta merancang data center yang lebih efisien dalam konsumsi listrik dan air.
- Merancang model AI yang task-specific (fokus satu tugas) karena dianggap lebih ramah lingkungan dibanding model AI multifungsi yang butuh konsumsi energi jauh lebih tinggi.
- Sebagai pengguna biasa, kita bisa berkontribusi dengan tidak asal-asalan memakai AI untuk hal sepele, karena setiap penggunaan tetap memakan energi di baliknya.
